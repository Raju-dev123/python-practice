problem 2 : word frequency counter

```python
#problem 2: word frequency counter
#problem statement

#take a sentence as input.

#display how many times each word appears.

#Eg1 : python is fun python is easy
#op:
#python - 2 , is - 2, easy - 1, fun - 1
```

```python
s = input()
words = s.split(" ")
print(words)
```

```python
freq = {}
for word in words:
  if word not in freq:
    freq[word] = 1
  else:
    freq[word] += 1

print(freq)
```

Problem 4 : Character counter

```python
problem statement

take a string as input from the user.

display

Number of vowels
Number of consonants
Number of digits
Number of spaces
```

```python
#Ip1: python3 is fun
#op1: vowels - 3
#Consonants - 8
#digits - 1
#spaces - 2
```

```python
s = input()
s1 = s.lower()

vowels = 0
consonants = 0
digits = 0
spaces = 0

for c in s1:
  if c in "aeiou":
    vowles += 1

  elif c >= "a" and c <= "z":
    consonants += 1
  elif c >= "0" and c <= "9":
    digits += 1
  elif c == " ":
    spaces += 1

print("Vowels -", vowels)
print("Consonants _", consonants)
print("Digits -", digits)
print("Spaces -", spaces)
```

```python
#version2

text = input()

vowels = 0
consonants = 0
digits = 0
spaces = 0

for ch in text:
  if ch.lower() in "aeiou":
    vowels += 1

  elif ch.isalpha():
    consonants += 1

  elif ch.isdigit():
    digits += 1

  elif ch.isspace():
    spaces += 1

print("Vowels -", vowels)
print("Consonants -", consonants)
print("Digits -", digits)
print("Spaces -", spaces)
```

```python
problem 6 : password validator
problem statement

A password is valid if it contains

minimum 8 characters
at least one uppercase letter
at least one lowercase letter
at least one digite

Display

Strong password

or

Weak password
```

```python
password = "Aditya789"

upper = False
lower = False
digit = False
n = len(password)

for c in password:
  if c.isupper():
    upper = True
  elif c.islower():
    lower = True
  elif c.isdigit():
    digit = True

if n >= 8 and upper==True and lower==True and digit==True:
  print("Strong password")
else:
  print("Weak password")
```
