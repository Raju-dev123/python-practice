```python
numbers = [1,2,3,4]
n = len(numbers)

for i in range(n):
  numbers[i] = numbers[i] + i

print(numbers)
```

```python
marks = [65, 75, 85]

total = 0
for m in marks:
  if m >= 70:
    total += m
print(total)
```

```python
names = ["Amit" , "Neha" , "Aditya"]
longest = names[0]
for name in names:
  if len(name) > len(longest):
    longest = name
print(longest)
```
