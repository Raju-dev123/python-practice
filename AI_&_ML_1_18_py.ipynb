Problem 7: Cricket Scoreboard

```python
problem statement

take player names and runs for 5 players.

store them in a dictionary.

Display

Highest scorer
Average scorer
```

```python
scores = {
    "Virat" : 90,
    "Rohit" : 97,
    "Dhoni" : 100,
    "Hardik" : 50,
    "Bhuvi" : 20
}
type(scores)
```

```python
mx_score = 0
total_score = 0
for key in scores:
  value = scores[key]

  if value > mx_score:
    mx_score = value
    mx_player = key

  total_score += value

print(mx_score)
print(mx_player)
print(total_score/len(scores))
```

problem 9: Most frequent character

```python
problem statement

find the charecter that appears the maximum number of time  in a string.

ignore spaces.
```

```python
s = "Python is the most amazing programing language for ML"
freq = {}

for c in s:
  if c != " ":
    if c not in freq:
      freq[c] = 1
    else:
      freq[c] += 1

print(freq)
```

```python
# __init__()
# __init__() ---> is built- in method which is alwayes executed when the class gets inited(object is beited)
```

```python
class person:
  def __init__(self,name,city):
    self.name = name
    self.city = city
```

```python
p1 = person("Raj","Mumbai")
print(p1.name, p1.city)
```

```python
p2 = person("Raju","Guwahati")
print(p2.name, p2.city)
```
